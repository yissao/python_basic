```python
def find_lines_with_string(filename, search_string, case_sensitive=True):
    """
    Finds and returns a list of lines from a file that contain the specified string.

    Args:
        filename (str): The path to the text file.
        search_string (str): The string to search for within each line.
        case_sensitive (bool): If True, the search is case-sensitive.
                               If False, the search is case-insensitive.

    Returns:
        list: A list of lines containing the search_string.
    """
    matching_lines = []
    try:
        with open(filename, 'r') as file:
            for line in file:
                if not case_sensitive:
                    if search_string.lower() in line.lower():
                        matching_lines.append(line.strip())  # .strip() removes leading/trailing whitespace including newline
                else:
                    if search_string in line:
                        matching_lines.append(line.strip())
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    return matching_lines

# Example usage:
arquivo_destino = "junta_linhas_01_01destino.txt"
file_path = "junta_linhas_01_01origem.txt"# example.txt"  # Replace with your file path
string_to_find = "feather"
sem_as_linhas = "junta_linhas_01_01sem_as_linhas.txt"
"""
# Create a dummy file for demonstration
with open(file_path, 'w') as f:
    f.write("This is a line about Python programming.\n")
    f.write("Another line, without the word.\n")
    f.write("python is versatile.\n")
    f.write("Final line with Python.\n")
"""

# Case-sensitive search
print("Case-sensitive search:")
found_lines_sensitive = find_lines_with_string(file_path, string_to_find, case_sensitive=True)
with open(arquivo_destino, 'w') as f:
    for line in found_lines_sensitive:
        f.write(line+'\n')

with open(sem_as_linhas, 'w') as g:
    if line.strip("\n") != string_to_find:
                g.write(line)

print("\nCase-insensitive search:")
# Case-insensitive search
found_lines_insensitive = find_lines_with_string(file_path, string_to_find, case_sensitive=False)
for line in found_lines_insensitive:
    print(line)

```
