```python
import re

# Caminhos dos arquivos
arquivo_entrada = 'seuarquivo.txt'
arquivo_saida = 'seuarquivo_formatado.txt'

# Lê o conteúdo do arquivo original
with open(arquivo_entrada, 'r', encoding='utf-8') as file:
    texto = file.read()

# Substituições com expressões regulares
texto = re.sub(r'•\n\n', '* ', texto)
texto = re.sub(r'◦\n\n', '\t- ', texto)
texto = re.sub(r'(\/\d\d) teste', r'\1 apagar ', texto)

# Grava o conteúdo modificado em um novo arquivo
with open(arquivo_saida, 'w', encoding='utf-8') as file:
    file.write(texto)

print(f"Substituições concluídas. Arquivo salvo como '{arquivo_saida}'.")
```
